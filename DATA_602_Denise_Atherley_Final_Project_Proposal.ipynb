**5. EDA and Summary Statistics:**

The exploratory data analysis began by loading the dataset and reviewing its overall structure. Initial inspection confirmed that the data contains a complete two-year record of daily observations with no missing values. The date field was converted from text format into a proper date variable, allowing the data to be organized chronologically and supporting future time-series analysis, seasonal comparisons, and trend visualization.

Several data preparation steps were performed to improve interpretability. Weather-related variables that were originally stored as normalized values were converted back to their real-world units, making temperature, humidity, and windspeed easier to understand and analyze. Additional descriptive variables were created by translating coded values into meaningful categories, such as season and weather condition, and by identifying whether each observation occurred on a working day or a weekend/holiday. These transformations expanded the dataset and provided more intuitive variables for analysis and modeling.

Summary statistics showed that overall daily bike ridership follows a roughly normal distribution, with the mean and median being very similar. This suggests that ridership levels are generally balanced and not heavily influenced by extreme values. When ridership was separated by user type, distinct patterns emerged. Registered users accounted for the majority of daily trips and displayed relatively consistent usage patterns, likely reflecting routine commuting behavior. In contrast, casual riders exhibited greater variability, with usage tending to increase on favorable weather days, weekends, and holidays.

Analysis of environmental variables indicated that the dataset captures a wide range of weather conditions, from cold winter days to warm summer periods. Temperature values were well distributed across seasons, while windspeed showed occasional higher values associated with stormier conditions. A review of humidity statistics also identified a potential data-quality issue, as one observation recorded a humidity value of zero percent, which is not physically realistic. This anomaly likely reflects a sensor or recording error and may require further treatment during the modeling phase. Overall, the dataset appears complete, well structured, and suitable for statistical analysis and predictive modeling.

```python
import pandas as pd
import numpy as np

# Load the dataset directly from the raw GitHub URL
url = "https://raw.githubusercontent.com/DRA-SPS27/DATA602-Final_Project/refs/heads/main/day.csv"
df = pd.read_csv(url)

# Verify dataset dimensions
print(f"Dataset shape: {df.shape}")

# Inspect the first 5 rows
print(df.head())

# 2. DATA WRANGLING
# Convert date string to datetime format
df["dteday"] = pd.to_datetime(df["dteday"])

# De-normalize weather variables into human-readable units for clear interpretation
df["temp_celsius"] = df["temp"] * 41.0
df["humidity_pct"] = df["hum"] * 100.0
df["windspeed_kmh"] = df["windspeed"] * 67.0

# Map categorical codes to descriptive text labels
season_map = {1: "Spring", 2: "Summer", 3: "Fall", 4: "Winter"}
weather_map = {
    1: "Clear / Partly Cloudy",
    2: "Mist / Cloudy",
    3: "Light Rain / Snow"
}
df["season_label"] = df["season"].map(season_map)
df["weather_label"] = df["weathersit"].map(weather_map)
df["day_type"] = np.where(df["workingday"] == 1, "Working Day", "Weekend / Holiday")

# Summary statistics for Proposal & EDA section
summary_cols = ["cnt", "casual", "registered", "temp_celsius", "humidity_pct", "windspeed_kmh"]
print("\nSummary Statistics:\n", df[summary_cols].describe().round(2))
```
